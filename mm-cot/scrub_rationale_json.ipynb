```python
import json
```

```python
with open('experiments/rationale_scrubbed_QCM-E/predictions_ans_test.json', 'r') as f:
    data = json.load(f)  # Correct way to load JSON from a file
```

```python
# Replace all "preds" fields in the JSON structure
# Function to recursively replace all preds list elements with "Solution:"
def replace_preds(obj):
    if isinstance(obj, dict):
        # If a "preds" key exists, replace each element in its list with "Solution:"
        if "preds" in obj:
            obj["preds"] = ["Solution:" for _ in obj["preds"]]
        # Recurse into each value
        for key in obj:
            replace_preds(obj[key])
    elif isinstance(obj, list):
        # Recurse into each item in the list
        for item in obj:
            replace_preds(item)
    return obj

# Apply replacement
data = replace_preds(data)

```

```python
# Save the modified JSON back to the file
with open('experiments/rationale_scrubbed_QCM-E/predictions_ans_test.json', "w") as f:
    json.dump(data, f, indent=4)
```
